## **Challenge #7: Two Lines**

Given:

days = [1, 2, 3, 4, 5]

taipei = [30, 31, 29, 32, 33] \
hualien = [28, 29, 28, 30, 31]

Plot both cities on the same graph.

Add a legend so you know which line is Taipei and which is Hualien.

```python
days = [1, 2, 3, 4, 5]

taipei = [30, 31, 29, 32, 33]
hualien = [28, 29, 28, 30, 31]

plt.plot(days, taipei, label='Taipei')
plt.plot(days, hualien, label='Hualien')
plt.title("Taipei vs Hualien Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature C")
plt.legend();
```

##**Mini project**

When those feel comfortable, try this without instructions:

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temperature = [26, 27, 29, 28, 30, 32, 31]\

Create a small weather report visualization where:

* temperature is a line graph
* rainfall is a bar chart
* both have titles
* both have axis labels
* the temperature graph has markers
* the temperature graph has a grid
* the graphs appear in the same figure

```python


days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temperature = [26, 27, 29, 28, 30, 32, 31]
rainfall = [10, 0, 5, 25, 15, 2, 8]

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(days, temperature, marker='o')
plt.grid()
plt.title("Temperature Forecast")
plt.xlabel('Days')
plt.ylabel('Temperature')

plt.subplot(1,2,2)
plt.bar(days, rainfall)
plt.title("Rainfall Forecast");
plt.xlabel("Days")
plt.ylabel("Rainfall");
```

## **experimenting with ax[]**

```python


days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temperature = [26, 27, 29, 28, 30, 32, 31]
rainfall = [10, 0, 5, 25, 15, 2, 8]

plt.figure(figsize=(10,4))

fig, ax = plt.subplots(1,2, figsize=(10,4))

ax[0].plot(days, temperature)
ax[0].grid()
ax[0].set_title("Temperature Forecast")
ax[0].set_xlabel('Days')
ax[0].set_ylabel('Temperature');


ax[1].bar(days, rainfall)
ax[1].set_title("Rainfall Forecast")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Rainfall");
plt.tight_layout()
```
